# Hash Map + Complement Pattern

The idea is to create a hash map to keep track if we have already seen the number that would "complement" our sum to reach the target.

For example:

target = 3

hash_map = {0: 1}

array = [1,1,1,2,3] 

prefix_sum = 0

count = 0

[1] add 1 to the prefix_sum (prefix_sum=1) and check if the (prefix_sum - target) has been seen before. Since -2 is not in the map (only 0 in the map so far), we just add the prefix_sum 1 to the map without counting anything.

[1] one more 1 to sum (prefix_sum=2) , prefix_sum - target = -1 , same as before, we dont have -1 in the map, we add the prefix_sum (2) to the map.

[1] again another 1 so that prefix_sum = 3, but now we have prefix_sum - target = 0, and we do have 0 in the map, meaning the sum so far complements the target. we count += 1 and add 3 (the sum so far) to the map.

[2] now we have prefix_sum = 5 and prefix_sum - target = 2. We have seen 2 before (in the map from step 2), so count+= 1 and we add 5 to the map (our sum so far).

[3] lastly, we have prefix_sum = 8 and prefix_sum - target = 5. We have 5 in the map, so we count += 1.   

### Two sums, LeetCode #1 | Easy

Given an array of integers nums and an integer target, return indices of the two numbers such that they add up to target.
You may assume that each input would have exactly one solution, and you may not use the same element twice.

```python
def twoSum(nums, target):
    
    hashmap = {}  # value: index
    
    for i, num in enumerate(nums):
        complement = target - num
    
        if complement in hashmap:
            return [hashmap[complement], i]
    
        hashmap[num] = i
```

### Count Binary Subarrays With Sum, LeetCode #930 | Medium
Given a binary array nums (only 0s and 1s), and an integer goal, return the number of non-empty subarrays that sum to goal.

```python
nums = [1,1,1,4,2,3]
#nums = [1,1,1,0,0,1]
k = 3

count = 0
prefix_map = {0:1}
prefix_sum = 0

for num in nums:
    prefix_sum += num
    complement = prefix_sum - k
    count += prefix_map.get(complement, 0)

    print(num, count)

    prefix_map[prefix_sum] = prefix_map.get(prefix_sum, 0) + 1

count
```

### Number of Subarrays with Sum Divisible by K, LeetCode #974 | Medium
Given an integer array nums and an integer k, return the number of subarrays whose sum is divisible by k.

```python
nums = [4, 5, 0, -2, -3, 1]
k = 5

prefix_sum = 0
prefix_map = {0: 1}
count = 0

for num in nums:
    prefix_sum += num

    remainder = (prefix_sum % k + k) % k # to make it positive
    count += prefix_map.get(remainder, 0)

    prefix_map[remainder] = prefix_map.get(remainder, 0) + 1

count
```
